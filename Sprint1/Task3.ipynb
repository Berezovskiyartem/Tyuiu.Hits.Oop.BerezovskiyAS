### Вариант задания 2


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Student в C#, который будет представлять информацию о
студентах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой

<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

// Базовый класс Student
public class Student
{
    // Приватные поля (инкапсуляция)
    private string _name;
    private int _age;
    private int _course;

    // Свойства с геттерами и сеттерами (с валидацией данных)
    public string Name
    {
        get => _name;
        set => _name = string.IsNullOrWhiteSpace(value) ? "Без имени" : value;
    }

    public int Age
    {
        get => _age;
        set => _age = value > 16 ? value : 16; // минимальный возраст 16
    }

    public int Course
    {
        get => _course;
        set => _course = (value >= 1 && value <= 6) ? value : 1;
    }

    // Конструктор
    public Student(string name, int age, int course)
    {
        Name = name;
        Age = age;
        Course = course;
    }

    // Методы базового класса
    public virtual string GetInfo()
    {
        return $"Студент: {Name}, Возраст: {Age}, Курс: {Course}";
    }

    public virtual void Study()
    {
        Console.WriteLine($"{Name} учится на {Course} курсе.");
    }

    public virtual void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен.");
    }

    // Взаимодействие объектов: совместная подготовка к экзамену
    public void PrepareForExamWith(Student partner)
    {
        Console.WriteLine($"[Взаимодействие] {Name} готовится к экзаменам вместе со студентом {partner.Name}.");
    }
}

// 1. Студент бакалавриата (BachelorStudent)
public class BachelorStudent : Student
{
    private string _specialty;

    public string Specialty
    {
        get => _specialty;
        set => _specialty = string.IsNullOrWhiteSpace(value) ? "Общий профиль" : value;
    }

    public BachelorStudent(string name, int age, int course, string specialty)
        : base(name, age, course)
    {
        Specialty = specialty;
    }

    // Переопределение метода Study()
    public override void Study()
    {
        Console.WriteLine($"{Name} учится на {Course} курсе по специальности «{Specialty}».");
    }

    // Взаимодействие объектов: взаимопомощь по специальности
    public void HelpWithSpecialty(BachelorStudent other)
    {
        Console.WriteLine($"[Взаимодействие] Бакалавр {Name} помогает бакалавру {other.Name} освоить предмет по специальности «{Specialty}».");
    }
}

// 2. Студент магистратуры (MasterStudent)
public class MasterStudent : Student
{
    private string _scientificAdvisor;

    public string ScientificAdvisor
    {
        get => _scientificAdvisor;
        set => _scientificAdvisor = string.IsNullOrWhiteSpace(value) ? "Не назначен" : value;
    }

    public MasterStudent(string name, int age, int course, string scientificAdvisor)
        : base(name, age, course)
    {
        ScientificAdvisor = scientificAdvisor;
    }

    // Переопределение метода TakeExam()
    public override void TakeExam()
    {
        Console.WriteLine($"{Name} сдает экзамен научному руководителю ({ScientificAdvisor}).");
    }
}

// 3. Аспирант (PhDStudent)
public class PhDStudent : Student
{
    private string _dissertationTopic;

    public string DissertationTopic
    {
        get => _dissertationTopic;
        set => _dissertationTopic = string.IsNullOrWhiteSpace(value) ? "Тема не утверждена" : value;
    }

    public PhDStudent(string name, int age, int course, string dissertationTopic)
        : base(name, age, course)
    {
        DissertationTopic = dissertationTopic;
    }

    // Переопределение метода Study()
    public override void Study()
    {
        Console.WriteLine($"{Name} изучает материалы по теме диссертации: «{DissertationTopic}».");
    }

    // Взаимодействие объектов: консультация младших студентов
    public void ConsultStudent(Student student)
    {
        Console.WriteLine($"[Взаимодействие] Аспирант {Name} консультирует студента {student.Name} по научной работе.");
    }
}

// Проверка работы программы
class Program
{
    static void Main()
    {
        // 1. Создание объектов через конструкторы
        BachelorStudent bachelor = new BachelorStudent("Артём", 20, 3, "Программная инженерия");
        MasterStudent master = new MasterStudent("Иван", 23, 1, "Проф. Иванов И. И.");
        PhDStudent phd = new PhDStudent("Елена", 26, 2, "Разработка систем на базе ИИ");

        Console.WriteLine("=== Информация о студентах ===");
        Console.WriteLine(bachelor.GetInfo());
        Console.WriteLine(master.GetInfo());
        Console.WriteLine(phd.GetInfo());

        Console.WriteLine("\n=== Демонстрация работы переопределенных методов ===");
        bachelor.Study();
        master.TakeExam();
        phd.Study();

        Console.WriteLine("\n=== Демонстрация взаимодействия объектов ===");
        // Взаимодействие между базовыми студентами
        bachelor.PrepareForExamWith(master);

        // Взаимодействие аспиранта с бакалавром
        phd.ConsultStudent(bachelor);

        // Взаимодействие двух бакалавров
        BachelorStudent bachelor2 = new BachelorStudent("Олег", 19, 2, "Программная инженерия");
        bachelor.HelpWithSpecialty(bachelor2); 
    }
}
```
