```
using System;
using System.Collections.Generic;
using System.Linq;

public enum SupplierType
{
    Logistics,
    OfficeEquipment,
    OfficeFurniture,
    IT,
    Other
}

public interface ISupplier
{
    string GetName();
    decimal GetCost();
    SupplierType Type { get; }
}

public interface IEmployee
{
    string GetFullName();
    decimal GetSalary();
    string Position { get; }
}

public interface IEnterprise
{
    void AddSupplier(ISupplier supplier);
    void AddEmployee(IEmployee employee);
    List<ISupplier> GetSuppliers();
    decimal CalculateTotalExpenses();
}

public class Supplier : ISupplier
{
    private readonly string _name;
    private readonly decimal _cost;

    public SupplierType Type { get; }

    public Supplier(string name, decimal cost, SupplierType type)
    {
        if (string.IsNullOrWhiteSpace(name))
            throw new ArgumentException("Название поставщика не может быть пустым", nameof(name));
        if (cost < 0)
            throw new ArgumentOutOfRangeException(nameof(cost), "Стоимость не может быть отрицательной");

        _name = name;
        _cost = cost;
        Type = type;
    }

    public string GetName() => _name;
    public decimal GetCost() => _cost;

    public override string ToString() =>
        $"{_name} – Тип: {Type} – Стоимость услуг: {_cost}";
}

public class Employee : IEmployee
{
    private readonly string _fullName;
    private readonly decimal _salary;

    public string Position { get; }

    public Employee(string fullName, string position, decimal salary)
    {
        if (string.IsNullOrWhiteSpace(fullName))
            throw new ArgumentException("Имя сотрудника не может быть пустым", nameof(fullName));
        if (string.IsNullOrWhiteSpace(position))
            throw new ArgumentException("Должность не может быть пустой", nameof(position));
        if (salary < 0)
            throw new ArgumentOutOfRangeException(nameof(salary), "Зарплата не может быть отрицательной");

        _fullName = fullName;
        Position = position;
        _salary = salary;
    }

    public string GetFullName() => _fullName;
    public decimal GetSalary() => _salary;

    public override string ToString() =>
        $"{_fullName} – Должность: {Position} – Зарплата: {_salary}";
}

public class Enterprise : IEnterprise
{
    private readonly List<ISupplier> _suppliers = new();
    private readonly List<IEmployee> _employees = new();

    public void AddSupplier(ISupplier supplier)
    {
        ArgumentNullException.ThrowIfNull(supplier);
        if (!_suppliers.Contains(supplier))
            _suppliers.Add(supplier);
    }

    public void AddEmployee(IEmployee employee)
    {
        ArgumentNullException.ThrowIfNull(employee);
        if (!_employees.Contains(employee))
            _employees.Add(employee);
    }

    public List<ISupplier> GetSuppliers() => new(_suppliers);

    public List<IEmployee> GetEmployees() => new(_employees);

    public decimal CalculateTotalExpenses()
    {
        decimal suppliersCost = _suppliers.Sum(s => s.GetCost());
        decimal employeesCost = _employees.Sum(e => e.GetSalary());
        return suppliersCost + employeesCost;
    }
}

// --- ИСПОЛНЯЕМЫЙ КОД (Top-level statements) ---

IEnterprise enterprise = new Enterprise();

enterprise.AddSupplier(new Supplier("Логистический центр", 100_000m, SupplierType.Logistics));
enterprise.AddSupplier(new Supplier("Поставщик офисной мебели", 50_000m, SupplierType.OfficeFurniture));
enterprise.AddSupplier(new Supplier("ИТ-оборудование", 75_000m, SupplierType.IT));

enterprise.AddEmployee(new Employee("Сергей Сергеев", "Менеджер", 70_000m));
enterprise.AddEmployee(new Employee("Евгения Смирнова", "Разработчик", 95_000m));

Console.WriteLine("Список поставщиков:");
var suppliers = enterprise.GetSuppliers();
for (int i = 0; i < suppliers.Count; i++)
{
    Console.WriteLine($"{i + 1}. {suppliers[i].GetName()} – " +
                      $"Тип: {suppliers[i].Type} – " +
                      $"Стоимость услуг: {suppliers[i].GetCost()}");
}

Console.WriteLine("\nСписок сотрудников:");
var employees = ((Enterprise)enterprise).GetEmployees();
for (int i = 0; i < employees.Count; i++)
{
    Console.WriteLine($"{i + 1}. {employees[i].GetFullName()} – " +
                      $"Должность: {employees[i].Position} – " +
                      $"Зарплата: {employees[i].GetSalary()}");
}

Console.WriteLine($"\nОбщие расходы: {enterprise.CalculateTotalExpenses()}");
```
