#### Дополнительное задание

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

public class Item
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Item(string name, decimal price)
    {
        Name = name;
        Price = price;
    }
}

public class Order
{
    public int OrderId { get; set; }
    public DateTime CreationDate { get; set; }
    public decimal TotalAmount { get; set; }
    public List<Item> Items { get; set; }
    public string Status { get; set; }
    public bool IsPaid { get; set; }

    public Order(int orderId, DateTime creationDate, string status, bool isPaid)
    {
        OrderId = orderId;
        CreationDate = creationDate;
        TotalAmount = 0m;
        Items = new List<Item>();
        Status = status;
        IsPaid = isPaid;
    }

    public virtual decimal CalculateTotal()
    {
        decimal sum = 0;
        foreach (var item in Items)
        {
            sum += item.Price;
        }
        TotalAmount = sum;
        return TotalAmount;
    }

    public virtual void AddItem(Item item)
    {
        Items.Add(item);
        CalculateTotal();
        Console.WriteLine($"Товар '{item.Name}' добавлен в заказ #{OrderId}.");
    }

    public virtual void RemoveItem(Item item)
    {
        if (Items.Contains(item))
        {
            Items.Remove(item);
            CalculateTotal();
            Console.WriteLine($"Товар '{item.Name}' удален из заказа #{OrderId}.");
        }
    }

    public void ApplyDiscount(decimal percent)
    {
        if (percent > 0 && percent <= 100)
        {
            TotalAmount -= TotalAmount * (percent / 100m);
            Console.WriteLine($"Применена скидка {percent}%. Итоговая сумма: {TotalAmount:C}");
        }
    }

    public void ApplyDiscount(decimal fixedAmount, string promoCode)
    {
        if (fixedAmount > 0 && fixedAmount <= TotalAmount)
        {
            TotalAmount -= fixedAmount;
            Console.WriteLine($"Применен промокод '{promoCode}'. Скидка: {fixedAmount:C}. Итого: {TotalAmount:C}");
        }
    }
}

public class OnlineOrder : Order
{
    public string CustomerEmail { get; set; }
    public string DeliveryService { get; set; }
    public string TrackingNumber { get; set; }

    public OnlineOrder(int orderId, DateTime creationDate, string status, bool isPaid, string customerEmail, string deliveryService, string trackingNumber)
        : base(orderId, creationDate, status, isPaid)
    {
        CustomerEmail = customerEmail;
        DeliveryService = deliveryService;
        TrackingNumber = trackingNumber;
    }

    public override void AddItem(Item item)
    {
        base.AddItem(item);
        Console.WriteLine($"Способ доставки для онлайн-заказа: курьерская служба {DeliveryService} (уведомление отправлено на {CustomerEmail}).");
    }

    public void SendEmailNotification()
    {
        Console.WriteLine($"Электронное письмо с трек-номером {TrackingNumber} отправлено на адрес {CustomerEmail}.");
    }
}

public class PhysicalOrder : Order
{
    public string DeliveryAddress { get; set; }
    public string StoreBranch { get; set; }
    public bool IsPickup { get; set; }

    public PhysicalOrder(int orderId, DateTime creationDate, string status, bool isPaid, string deliveryAddress, string storeBranch, bool isPickup)
        : base(orderId, creationDate, status, isPaid)
    {
        DeliveryAddress = deliveryAddress;
        StoreBranch = storeBranch;
        IsPickup = isPickup;
    }

    public override void RemoveItem(Item item)
    {
        base.RemoveItem(item);
        Console.WriteLine($"Оформлен возврат товара '{item.Name}' на склад физического магазина '{StoreBranch}' по адресу: {DeliveryAddress}.");
    }

    public void UpdateDeliveryAddress(string newAddress)
    {
        DeliveryAddress = newAddress;
        Console.WriteLine($"Адрес доставки изменен на: {DeliveryAddress}");
    }
}

public class SpecializedOrder : Order
{
    public string SpecialConditions { get; set; }
    public decimal AdditionalFee { get; set; }
    public bool RequiresSpecialHandling { get; set; }

    public SpecializedOrder(int orderId, DateTime creationDate, string status, bool isPaid, string specialConditions, decimal additionalFee, bool requiresSpecialHandling)
        : base(orderId, creationDate, status, isPaid)
    {
        SpecialConditions = specialConditions;
        AdditionalFee = additionalFee;
        RequiresSpecialHandling = requiresSpecialHandling;
    }

    public override decimal CalculateTotal()
    {
        decimal baseSum = base.CalculateTotal();
        if (RequiresSpecialHandling)
        {
            TotalAmount = baseSum + AdditionalFee;
        }
        return TotalAmount;
    }

    public void PrintSpecialRequirements()
    {
        Console.WriteLine($"Специальные условия заказа: {SpecialConditions}. Доплата за обслуживание: {AdditionalFee:C}");
    }
}

public class OrderProcessor<T> where T : Order
{
    public T ProcessedOrder { get; set; }

    public OrderProcessor(T processedOrder)
    {
        ProcessedOrder = processedOrder;
    }

    public void ExecuteProcessing()
    {
        Console.WriteLine($"Обработка заказа ID: {ProcessedOrder.OrderId}. Текущая сумма: {ProcessedOrder.CalculateTotal():C}");
    }
}

Item item1 = new Item("Ноутбук", 80000m);
Item item2 = new Item("Мышь", 3000m);

OnlineOrder onlineOrder = new OnlineOrder(101, DateTime.Now, "Новый", true, "artem@mail.ru", "СДЭК", "TRK-998877");
onlineOrder.AddItem(item1);
onlineOrder.SendEmailNotification();
Console.WriteLine();

PhysicalOrder physicalOrder = new PhysicalOrder(102, DateTime.Now, "В обработке", false, "ул. Ленина, 10", "Тюмень Молл", false);
physicalOrder.AddItem(item2);
physicalOrder.RemoveItem(item2);
Console.WriteLine();

SpecializedOrder specOrder = new SpecializedOrder(103, DateTime.Now, "Срочный", true, "Хрупкий груз", 1500m, true);
specOrder.AddItem(item1);
specOrder.PrintSpecialRequirements();
Console.WriteLine($"Итоговая сумма с учетом условий: {specOrder.CalculateTotal():C}");
Console.WriteLine();

OrderProcessor<OnlineOrder> processor = new OrderProcessor<OnlineOrder>(onlineOrder);
processor.ExecuteProcessing();
```
