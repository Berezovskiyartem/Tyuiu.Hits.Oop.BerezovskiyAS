```
using System;

public interface IBonusProgram
{
    int BonusPoints { get; set; }
    void AccumulateBonuses(decimal amount);
}

public interface ISecureable
{
    string PinCode { get; set; }
    void ChangePin(string newPin);
}

public class CreditCard
{
    public string CardNumber { get; set; }
    public string HolderName { get; set; }
    public string ExpiryDate { get; set; }
    public decimal Balance { get; set; }
    public string Currency { get; set; }
    public bool IsActive { get; set; }

    public CreditCard(string cardNumber, string holderName, string expiryDate, decimal balance, string currency, bool isActive)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        Balance = balance;
        Currency = currency;
        IsActive = isActive;
    }

    public virtual string GetInfo()
    {
        return $"Карта: {CardNumber}, Владелец: {HolderName}, Срок: {ExpiryDate}, Баланс: {Balance} {Currency}";
    }

    public virtual void Pay(decimal amount)
    {
        if (IsActive && amount > 0 && Balance >= amount)
        {
            Balance -= amount;
            Console.WriteLine($"Успешно оплачено {amount} {Currency} с карты {CardNumber}.");
        }
        else
        {
            Console.WriteLine($"Ошибка оплаты с карты {CardNumber}.");
        }
    }

    public virtual void CheckBalance()
    {
        Console.WriteLine($"Текущий баланс карты {CardNumber}: {Balance} {Currency}");
    }

    public void BlockCard()
    {
        IsActive = false;
        Console.WriteLine($"Карта {CardNumber} заблокирована.");
    }
}

public class GoldCard : CreditCard, IBonusProgram, ISecureable
{
    public int BonusMiles { get; set; }
    public int BonusPoints { get; set; }
    public string PinCode { get; set; }
    public double CashBackRate { get; set; }

    public GoldCard(string cardNumber, string holderName, string expiryDate, decimal balance, string currency, bool isActive,
                    int bonusMiles, int bonusPoints, string pinCode, double cashBackRate)
        : base(cardNumber, holderName, expiryDate, balance, currency, isActive)
    {
        BonusMiles = bonusMiles;
        BonusPoints = bonusPoints;
        PinCode = pinCode;
        CashBackRate = cashBackRate;
    }

    public override void Pay(decimal amount)
    {
        base.Pay(amount);
        if (amount > 0)
        {
            int earnedMiles = (int)(amount / 100);
            BonusMiles += earnedMiles;
            AccumulateBonuses(amount);
            Console.WriteLine($"Начислено миль: {earnedMiles}. Всего миль: {BonusMiles}");
        }
    }

    public void AccumulateBonuses(decimal amount)
    {
        int points = (int)(amount * 0.05m);
        BonusPoints += points;
        Console.WriteLine($"Начислено бонусных баллов: {points}. Всего баллов: {BonusPoints}");
    }

    public void ChangePin(string newPin)
    {
        PinCode = newPin;
        Console.WriteLine("ПИН-код для GoldCard успешно изменен.");
    }

    public void TransferBonusToMiles()
    {
        BonusMiles += BonusPoints / 10;
        BonusPoints = 0;
        Console.WriteLine($"Бонусные баллы конвертированы в мили. Миль: {BonusMiles}");
    }
}

public class PremiumCard : CreditCard
{
    public string SupportAssistant { get; set; }
    public decimal CreditLimit { get; set; }
    public int FreeAirportLounges { get; set; }

    public PremiumCard(string cardNumber, string holderName, string expiryDate, decimal balance, string currency, bool isActive,
                       string supportAssistant, decimal creditLimit, int freeAirportLounges)
        : base(cardNumber, holderName, expiryDate, balance, currency, isActive)
    {
        SupportAssistant = supportAssistant;
        CreditLimit = creditLimit;
        FreeAirportLounges = freeAirportLounges;
    }

    public override void CheckBalance()
    {
        base.CheckBalance();
        Console.WriteLine($"Персональный ассистент {SupportAssistant} на связи. Доступный кредитный лимит: {CreditLimit} {Currency}");
    }

    public void RequestConciergeService(string requestText)
    {
        Console.WriteLine($"Ассистент {SupportAssistant} обрабатывает ваш запрос: '{requestText}'");
    }

    public void UseLoungeAccess()
    {
        if (FreeAirportLounges > 0)
        {
            FreeAirportLounges--;
            Console.WriteLine($"Проход в бизнес-зал оформлен. Осталось проходов: {FreeAirportLounges}");
        }
        else
        {
            Console.WriteLine("Бесплатные проходы в бизнес-зал исчерпаны.");
        }
    }
}

public class CorporateCard : CreditCard
{
    public string Company { get; set; }
    public string TaxNumber { get; set; }
    public decimal MonthlyExpenseLimit { get; set; }

    public CorporateCard(string cardNumber, string holderName, string expiryDate, decimal balance, string currency, bool isActive,
                         string company, string taxNumber, decimal monthlyExpenseLimit)
        : base(cardNumber, holderName, expiryDate, balance, currency, isActive)
    {
        Company = company;
        TaxNumber = taxNumber;
        MonthlyExpenseLimit = monthlyExpenseLimit;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Компания: {Company}, ИНН: {TaxNumber}";
    }

    public void GenerateExpenseReport()
    {
        Console.WriteLine($"Сформирован отчет о расходах для компании {Company} (ИНН: {TaxNumber}).");
    }

    public void UpdateExpenseLimit(decimal newLimit)
    {
        MonthlyExpenseLimit = newLimit;
        Console.WriteLine($"Лимит расходов для карты изменен на {MonthlyExpenseLimit} {Currency}.");
    }
}

public class VIPCorporateCard : CorporateCard
{
    public string AccountManager { get; set; }
    public bool InternationalInsurance { get; set; }
    public decimal OverdraftLimit { get; set; }

    public VIPCorporateCard(string cardNumber, string holderName, string expiryDate, decimal balance, string currency, bool isActive,
                            string company, string taxNumber, decimal monthlyExpenseLimit,
                            string accountManager, bool internationalInsurance, decimal overdraftLimit)
        : base(cardNumber, holderName, expiryDate, balance, currency, isActive, company, taxNumber, monthlyExpenseLimit)
    {
        AccountManager = accountManager;
        InternationalInsurance = internationalInsurance;
        OverdraftLimit = overdraftLimit;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", VIP-менеджер: {AccountManager}, Овердрафт: {OverdraftLimit} {Currency}";
    }

    public void ContactManager()
    {
        Console.WriteLine($"Связь с VIP-менеджером {AccountManager} установлена.");
    }

    public void CheckInsuranceStatus()
    {
        Console.WriteLine(InternationalInsurance ? "Международная страховка активна." : "Страховка не оформлена.");
    }
}

GoldCard goldCard = new GoldCard("4000 1234 5678 9010", "Артём", "12/28", 50000m, "RUB", true, 120, 500, "1234", 1.5);
PremiumCard premiumCard = new PremiumCard("4000 9876 5432 1090", "Артём", "05/29", 150000m, "RUB", true, "Елена", 500000m, 4);
VIPCorporateCard vipCard = new VIPCorporateCard("4000 5555 4444 3333", "Артём", "10/27", 300000m, "RUB", true, "ТехноПром", "7203123456", 1000000m, "Максим", true, 200000m);

goldCard.Pay(10000m);
goldCard.ChangePin("5678");
Console.WriteLine();

premiumCard.CheckBalance();
premiumCard.RequestConciergeService("Забронировать отель");
Console.WriteLine();

Console.WriteLine(vipCard.GetInfo());
vipCard.ContactManager();
```
