<h2 style="color:DodgerBlue">Описание проекта:</h2>

Создать базовый класс Product в C#, который будет представлять информацию о
продуктах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;

public interface IRecyclable
{
    bool IsEcoFriendly { get; set; }
    void Recycle();
}

public interface IWarrantable
{
    int WarrantyPeriod { get; set; }
    void ExtendWarranty(int extraMonths);
}

public class Product
{
    public string Name { get; set; }
    public decimal Price { get; set; }
    public string Manufacturer { get; set; }
    public int InStockQuantity { get; set; }

    public Product(string name, decimal price, string manufacturer, int inStockQuantity)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        InStockQuantity = inStockQuantity;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Производитель: {Manufacturer}, Цена: {Price:C}, В наличии: {InStockQuantity} шт.";
    }

    public virtual void Discount(decimal percentage)
    {
        if (percentage > 0 && percentage <= 100)
        {
            Price -= Price * (percentage / 100);
        }
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }

    public void Restock(int quantity)
    {
        Console.WriteLine($"Пополнение склада для '{Name}': +{quantity} шт.");
        InStockQuantity += quantity;
    }
}

public class Electronics : Product, IRecyclable, IWarrantable
{
    public int PowerWatts { get; set; }
    public bool IsEcoFriendly { get; set; }
    public int WarrantyPeriod { get; set; }

    public Electronics(string name, decimal price, string manufacturer, int inStockQuantity, 
                       int powerWatts, bool isEcoFriendly, int warrantyPeriod)
        : base(name, price, manufacturer, inStockQuantity)
    {
        PowerWatts = powerWatts;
        IsEcoFriendly = isEcoFriendly;
        WarrantyPeriod = warrantyPeriod;
    }

    public override void Discount(decimal percentage)
    {
        decimal totalDiscount = percentage;
        if (WarrantyPeriod > 12)
        {
            totalDiscount += 5;
        }
        base.Discount(totalDiscount);
    }

    public void Recycle()
    {
        Console.WriteLine(IsEcoFriendly 
            ? $"Устройство '{Name}' отправлено на эко-переработку." 
            : $"Устройство '{Name}' требует специальной утилизации.");
    }

    public void ExtendWarranty(int extraMonths)
    {
        WarrantyPeriod += extraMonths;
        Console.WriteLine($"Гарантия на '{Name}' продлена на {extraMonths} мес.");
    }

    public void RunDiagnostics()
    {
        Console.WriteLine($"Диагностика '{Name}': Энергопотребление — {PowerWatts} Вт.");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Мощность: {PowerWatts} Вт, Гарантия: {WarrantyPeriod} мес.";
    }
}

public class Clothing : Product
{
    public string Size { get; set; }
    public string Material { get; set; }
    public string Gender { get; set; }

    public Clothing(string name, decimal price, string manufacturer, int inStockQuantity,
                    string size, string material, string gender)
        : base(name, price, manufacturer, inStockQuantity)
    {
        Size = size;
        Material = material;
        Gender = gender;
    }

    public override void Display()
    {
        Console.WriteLine($"{GetInfo()}, Размер: {Size}, Материал: {Material}, Категория: {Gender}");
    }
}

public class Books : Product
{
    public string Author { get; set; }
    public int PageCount { get; set; }
    public string Genre { get; set; }

    public Books(string name, decimal price, string manufacturer, int inStockQuantity,
                 string author, int pageCount, string genre)
        : base(name, price, manufacturer, inStockQuantity)
    {
        Author = author;
        PageCount = pageCount;
        Genre = genre;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Автор: {Author}, Жанр: {Genre}";
    }
}

public class SmartPhone : Electronics
{
    public string OperatingSystem { get; set; }
    public int BatteryCapacity { get; set; }
    public int CameraMegapixels { get; set; }

    public SmartPhone(string name, decimal price, string manufacturer, int inStockQuantity,
                      int powerWatts, bool isEcoFriendly, int warrantyPeriod,
                      string os, int batteryCapacity, int cameraMegapixels)
        : base(name, price, manufacturer, inStockQuantity, powerWatts, isEcoFriendly, warrantyPeriod)
    {
        OperatingSystem = os;
        BatteryCapacity = batteryCapacity;
        CameraMegapixels = cameraMegapixels;
    }

    public void TakePhoto()
    {
        Console.WriteLine($"Снимок сделан на камеру {CameraMegapixels} Мп.");
    }

    public override void Display()
    {
        Console.WriteLine($"{GetInfo()}, ОС: {OperatingSystem}, Батарея: {BatteryCapacity} мАч");
    }
}

Console.WriteLine("=== 1. Основное задание ===");
Electronics laptop = new Electronics("Ноутбук", 80000m, "ASUS", 5, 65, true, 24);
Clothing jacket = new Clothing("Куртка", 15000m, "Nike", 12, "L", "Нейлон", "Мужская");
Books book = new Books("Идиот", 900m, "Эксмо", 20, "Ф. М. Достоевский", 640, "Роман");

laptop.Display();
jacket.Display();
book.Display();

Console.WriteLine("\nПрименение скидки 10%:");
laptop.Discount(10);
jacket.Discount(10);
book.Discount(10);

laptop.Display();
jacket.Display();
book.Display();

Console.WriteLine("\n=== 2. Дополнительное задание ===");
SmartPhone phone = new SmartPhone("iPhone 15 Pro", 120000m, "Apple", 8, 15, true, 12, "iOS", 3274, 48);
phone.Display();
phone.RunDiagnostics();
phone.TakePhoto();
phone.ExtendWarranty(12);
phone.Recycle();
```
