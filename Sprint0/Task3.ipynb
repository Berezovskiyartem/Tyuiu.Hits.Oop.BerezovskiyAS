### Вариант задания 1


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Person в C#, который будет представлять информацию о
человеке. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.


<h2 style="color:DodgerBlue">Реализация:</h2>

----

```
using System;

// Базовый класс Person
public class Person
{
    // Атрибуты (свойства)
    public string Name { get; set; }
    public int Age { get; set; }
    public string Gender { get; set; }

    // Конструктор
    public Person(string name, int age, string gender) 
    {
        Name = name;
        Age = age;
        Gender = gender;
    }

    // Метод для получения информации о человеку
    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}";
    }

    // Метод для вывода приветствия
    public virtual void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}.");
    }
}

// 1. Производный класс Student
public class Student : Person
{
    public string University { get; set; }
    public int Course { get; set; }

    public Student(string name, int age, string gender, string university, int course)
        : base(name, age, gender)
    {
        University = university;
        Course = course;
    }

    // Переопределение метода SayHello() с добавлением информации о курсе
    public override void SayHello()
    {
        Console.WriteLine($"Привет! Меня зовут {Name}, я учусь на {Course} курсе в {University}.");
    }
}

// 2. Производный класс Employee
public class Employee : Person
{
    public string Company { get; set; }
    public decimal Salary { get; set; }

    public Employee(string name, int age, string gender, string company, decimal salary)
        : base(name, age, gender)
    {
        Company = company;
        Salary = salary;
    }

    // Переопределение метода GetInfo() для включения компании и зарплаты
    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Компания: {Company}, Зарплата: {Salary} руб.";
    }
}

// 3. Производный класс Teacher
public class Teacher : Person
{
    public string Subject { get; set; }
    public int Experience { get; set; }

    public Teacher(string name, int age, string gender, string subject, int experience)
        : base(name, age, gender)
    {
        Subject = subject;
        Experience = experience;
    }

    // Переопределение метода SayHello() с добавлением информации о предмете
    public override void SayHello()
    {
        Console.WriteLine($"Здравствуйте! Меня зовут {Name}, я преподаю {Subject} (стаж: {Experience} лет).");
    }
}

// Демонстрация работы (Полиморфизм)
class Program
{
    static void Main()
    {
        // Создание объектов
        Person person = new Person("Алексей", 30, "Мужской");
        Student student = new Student("Артём", 20, "Мужской", "ТИУ", 3);
        Employee employee = new Employee("Иван", 28, "Мужской", "Лаборатория Касперского", 150000);
        Teacher teacher = new Teacher("Елена Петровна", 45, "Женский", "Программирование C#", 15);

        // Массив объектов базового класса для демонстрации полиморфизма
        Person[] people = { person, student, employee, teacher };

        Console.WriteLine("--- Демонстрация методов GetInfo() ---");
        foreach (var p in people)
        {
            Console.WriteLine(p.GetInfo());
        }

        Console.WriteLine("\n--- Демонстрация методов SayHello() ---");
        foreach (var p in people)
        {
            p.SayHello();
        }
    }
}
```
